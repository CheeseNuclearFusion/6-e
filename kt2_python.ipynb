```python
import os
import random
import sys

def load_words():
    with open("slova.txt", "r", encoding="utf-8") as f:
        words = [line.strip().split("|") for line in f if "|" in line]
    random.shuffle(words)
    return words

def load_stage(errors):
    path = f"hangman_stages/stage_{min(errors, 6)}.txt"
    if os.path.exists(path):
        with open(path, "r", encoding="utf-8") as f:
            return f.read()
    return f"[Ошибок: {errors}]"

def draw(word_guess, errors, guessed, desc, max_err=6):
    os.system('cls' if os.name == 'nt' else 'clear')
    print(load_stage(errors))
    print(f"\n {desc}\n")
    print(" ".join(word_guess))
    print(f"\n Жизни: {max_err - errors}/{max_err}")
    if guessed:
        print(f" Использовано: {', '.join(sorted(guessed))}")

def play(word, desc):
    guessed, errors = set(), 0
    mask = ['■'] * len(word)
    max_err = 6

    while errors < max_err and '■' in mask:
        draw(mask, errors, guessed, desc)
        inp = input("\n Буква или слово: ").lower().strip()

        if not inp:
            continue

        if len(inp) > 1:
            if inp == word:
                print(f"\n ПОБЕДА! Слово: {word.upper()}")
                return True
            errors += 1
            continue

        if inp in guessed:
            print("Уже была такая буква!")
            continue

        guessed.add(inp)
        if inp in word:
            for i, ch in enumerate(word):
                if ch == inp:
                    mask[i] = ch
            print(f" Есть буква '{inp}'!")
        else:
            errors += 1
            print(f" Нет буквы '{inp}'!")

        if '■' not in mask:
            draw(mask, errors, guessed, desc)
            print(f"\n ПОБЕДА! Слово: {word.upper()}")
            return True

        input("\nEnter...")

    draw(mask, errors, guessed, desc)
    print(f"\n Вас повесили! Слово: {word.upper()}")
    return False

def main():
    words = load_words()
    wins = 0

    print("=" * 40)
    print("ВИСЕЛИЦА НА ПОЛЕ ЧУДЕС")
    print(f"Слов в базе: {len(words)}")
    print("=" * 40)
    input("Enter...")

    for i, (word, desc) in enumerate(words, 1):
        print(f"\n Раунд {i}/{len(words)} | Побед: {wins}")
        if play(word, desc):
            wins += 1

        if i < len(words):
            again = input("\nЕщё раунд? (да/нет): ").lower()
            if again not in ['да', 'yes', 'д', 'y']:
                break

    print(f"\n ИГРА ОКОНЧЕНА! Побед: {wins}/{i}")
    print(f"Процент: {wins/i*100:.1f}%")

if __name__ == "__main__":
    main()
```
