```python
import csv

# задание 1
def get_books(filename):
    books = []

    with open(filename, 'r', encoding='utf-8') as file:

        lines = file.readlines()

        for line in lines:
            line = line.strip()
            if not line:
                continue

            if line == 'isbn|title|author|quantity|price' or line.isdigit():
                continue

            parts = line.split('|')

            if len(parts) == 5:
                isbn = parts[0]
                title = parts[1]
                author = parts[2]
                quantity = int(parts[3])
                price = float(parts[4])

                books.append([isbn, title, author, quantity, price])

    return books


# задание 2
def filtered_books(books, search_string):
    result = []
    search_lower = search_string.lower()

    for book in books:
        title = book[1].lower()

        if search_lower in title:

            combined_title_author = f"{book[1]}, {book[2]}"
            result.append([book[0], combined_title_author, book[3], book[4]])

    return result


# задание 3
def calculate_total_cost(filtered_books):
    result = []

    for book in filtered_books:
        isbn = book[0]
        quantity = book[2]
        price = book[3]
        total_cost = quantity * price

        result.append((isbn, total_cost))

    return result
```
